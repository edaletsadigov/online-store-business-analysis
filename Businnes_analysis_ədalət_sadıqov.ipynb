## Metrics

### Product metrics

#### DAU WAU MAU

- `DAU` - Dayly active user
- `WAU` - Weeekly active user
- `MAU` - Montly active user

```python
visit['session_date'] = visit['start_ts'].dt.date
visit['session_week'] = visit['start_ts'].dt.to_period('W')
visit['session_month'] = visit['start_ts'].dt.to_period('M')
```

```python
dau = visit.groupby('session_date')['uid'].nunique().mean()
wau = visit.groupby('session_week')['uid'].nunique().mean()
mau = visit.groupby('session_month')['uid'].nunique().mean()


sticky_factor = dau/mau * 100
```

```python
print(f'Average DAU : {dau:.0f}')
print(f'Average WAU : {wau:.0f}')
print(f'Average MAU : {mau:.0f}')
print(f'Sticky factor (DAU/MAU): {sticky_factor:.1f}%')
```

```python
dau_daily = visit.groupby('session_date')['uid'].nunique().reset_index()
dau_daily.columns = ['date', 'dau']
```

```python
fig_daily = px.line(
    dau_daily,
    x='date',
    y='dau',
    title='Number of Daily Active Users',
    labels={
        'date': 'Date',
        'dau': 'Number of Unique Visitors'
    }
)

fig_daily.show()
```

**Result:** The average sticky factor (DAU/MAU) is approximately **4%**, meaning only 4% of monthly active users visit the site on a typical day. This indicates low daily user retention. The metric rises noticeably toward December, which may suggest **seasonal demand** related to the holiday period [not directly verified].

#### Retention rate

```python
first_activity = visit.groupby('uid')['session_month'].min().reset_index()
first_activity.columns = ['uid', 'first_activity_month']
```

```python
visit_cohort = visit.merge(first_activity, on='uid')
visit_cohort['cohort_age'] = (
    visit_cohort['session_month'] - visit_cohort['first_activity_month']
)
```

```python

cohort_sizes = visit_cohort.groupby('first_activity_month')['uid'].nunique()
cohort_sizes.name = 'n_users'
```

```python

visit_cohort = visit.merge(first_activity, on='uid')

visit_cohort['cohort_age'] = (
    (visit_cohort['session_month'].dt.year -
     visit_cohort['first_activity_month'].dt.year) * 12
    + visit_cohort['session_month'].dt.month
    - visit_cohort['first_activity_month'].dt.month
).astype('int64')
```

```python

cohort_activity = (
    visit_cohort
    .groupby(['first_activity_month', 'cohort_age'])['uid']
    .nunique()
    .reset_index(name='n_active')
)

cohort_activity = cohort_activity.merge(
    cohort_sizes.rename('n_users').reset_index(),
    on='first_activity_month',
    how='left',
    validate='many_to_one'
)

cohort_activity['retention'] = (
    cohort_activity['n_active'] / cohort_activity['n_users']
)
cohort_activity['retention'] = cohort_activity['n_active'] / cohort_activity['n_users']
```

```python

retention_pivot = cohort_activity.pivot_table(
    index='first_activity_month', columns='cohort_age', values='retention'
)
retention_pivot.round(3)
```

```python
retention_pivot_copy = retention_pivot.copy()
retention_pivot_copy.index = retention_pivot_copy.index.astype(str)
```

```python
retention_fig = px.imshow(
    retention_pivot_copy,
    text_auto='.0%',
    aspect='auto',
    color_continuous_scale='Blues',
    title='User Retention Matrix — Month × Cohort',
    labels={
        'x': 'Cohort ',
        'y': 'The first activity month (cohort)',
        'color': 'Retention'
    }
)

retention_fig.show()
```

Retention drops sharply to **~7% in Month 2** and stabilizes at **~3–5%** thereafter, indicating **low long-term retention** and predominantly event-specific, one-time usage.

#### LTV (Customer Lifetime Value)

```python
orders_ltv = orders.merge(first_visit,on='uid',how='left')
orders_ltv.sample(5)
```

```python
orders_ltv['cohort_month'] = orders_ltv['first_visit_ts'].dt.to_period('M')
orders_ltv['month_number'] = (orders_ltv['order_month']  - orders_ltv['cohort_month']).apply(lambda x: x.n)
```

```python
orders_ltv.sample(5)
```

```python
size_of_buyers_cohort = (
    first_visit.assign(cohort_month=first_visit['first_visit_ts'].dt.to_period('M'))
    .groupby('cohort_month')['uid']
    .nunique()
    .rename('n_buyers')
)

size_of_buyers_cohort
```

```python
revenue_by_month_num = orders_ltv.groupby(['cohort_month', 'month_number'])['revenue'].sum().reset_index()
revenue_by_month_num.sample(5)
```

```python
revenue_by_month_num = revenue_by_month_num.merge(size_of_buyers_cohort,on='cohort_month')
revenue_by_month_num.sample(5)
```

```python
revenue_by_month_num['lifetime_value'] = (revenue_by_month_num['revenue']/revenue_by_month_num['n_buyers']).round(5)

revenue_by_month_num.sample(5)
```

```python
ltv_pivot = revenue_by_month_num.pivot_table(index='cohort_month',columns='month_number',values='lifetime_value')
ltv_cumulative = ltv_pivot.cumsum(axis=1)

print(ltv_cumulative)
```

```python
ltv_heatmap = ltv_cumulative.copy().fillna(0)

ltv_heatmap.index = ltv_heatmap.index.astype(str)
ltv_heatmap.columns = [str(col) for col in ltv_heatmap.columns]

fig_heatmap = px.imshow(
    ltv_heatmap,
    text_auto=".2f",
    aspect="auto",
    color_continuous_scale="Blues",
    title="Cumulative LTV by Cohort and Month",
    labels={
        "x": "Month number",
        "y": "Cohort month",
        "color": "Cumulative LTV"
    }
)

fig_heatmap.update_layout(
    xaxis_title="Month number",
    yaxis_title="Cohort month",
    coloraxis_colorbar=dict(title="Cumulative LTV")
)

fig_heatmap.show()
```

## Additional

```python
daily_costs = cost.groupby('dt')['costs'].sum()
daily_revenue = orders.groupby(orders['buy_ts'].dt.date)['revenue'].sum()
daily_revenue.index = pd.to_datetime(daily_revenue.index)
```

```python

merged_daily = pd.concat([daily_costs.rename('cost'), daily_revenue.rename('revenue')], axis=1).dropna()

corr, p_val = st.spearmanr(merged_daily['cost'], merged_daily['revenue'])
print(f"Spearmen correlation value: {corr:.3f} (p-value={p_val:.4f})")
```

**Result:** Positive correlation between spend and revenue does not prove causation, as both may be driven by seasonal demand. **Controlled experiments** are needed to confirm the impact of marketing spend.
